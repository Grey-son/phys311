```python
import numpy as np
import matplotlib.pyplot as plt

# === Free-fall simulation using Euler's method ===
# A ball dropped from 100 m — your first numerical solve!

g = 9.8      # gravitational acceleration (m/s^2)
dt = 0.01    # time step (s)
t_max = 4.5  # total time (s)

# Initialize arrays
t = np.arange(0, t_max, dt)
y = np.zeros_like(t)       # position
v = np.zeros_like(t)       # velocity

# Initial conditions
y[0] = 100.0   # 100 meters above ground
v[0] = 0.0     # released from rest

# Euler integration
for i in range(len(t) - 1):
    v[i+1] = v[i] - g * dt
    y[i+1] = y[i] + v[i] * dt
    if y[i+1] < 0:          # stop at ground
        y[i+1] = 0
        v[i+1] = 0

# Exact solution for comparison
y_exact = 100 - 0.5 * g * t**2
y_exact = np.maximum(y_exact, 0)

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(t, y, 'o-', markersize=2, label='Euler (numerical)', color='#ff8200')
ax1.plot(t, y_exact, '--', label='Exact solution', color='#333333')
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Height (m)')
ax1.set_title('Free Fall: Position vs. Time')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(t, v, 'o-', markersize=2, color='#ff8200')
ax2.set_xlabel('Time (s)')
ax2.set_ylabel('Velocity (m/s)')
ax2.set_title('Free Fall: Velocity vs. Time')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('test_freefall.png', dpi=150)
plt.show()

print("\n✅ Setup verified! NumPy, Matplotlib, and Jupyter are working.")
print("   You just ran your first numerical physics simulation.")
print("   Welcome to Physics 311!")
```
